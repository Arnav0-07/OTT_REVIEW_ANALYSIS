```python
import csv

# --------------------------------------------
# OTT RECOMMENDATION SYSTEM
# Beginner-friendly Python mini-project
# Dataset: 100 sample titles
#created by ARNAV MISHRA (abes engineering collage)
# --------------------------------------------

file_name = "ott_dataset_100.csv"


def load_movies():
    movies = []

    try:
        with open(file_name, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                row["rating"] = float(row["rating"])
                movies.append(row)

    except FileNotFoundError:
        print("Dataset file not found!")
        print("Keep ott_dataset_100.csv in the same folder as this Python file.")

    return movies


def show_recommendations(movie_list):
    if len(movie_list) == 0:
        print("No matching titles found.")
    else:
        movie_list.sort(key=lambda movie: movie["rating"], reverse=True)

        print("\n------ YOUR RECOMMENDATIONS ------")

        for movie in movie_list[:5]:
            print("\nTitle:", movie["title"])
            print("Platform:", movie["platform"])
            print("Genre:", movie["genre"])
            print("Sample rating:", movie["rating"], "/ 5")
            print("Description:", movie["description"])


movies = load_movies()

if len(movies) == 0:
    print("Please check your dataset file.")
else:
    print("Welcome to the OTT Recommendation System!")
    print("Dataset loaded successfully:", len(movies), "titles")

    while True:
        print("\n========== MAIN MENU ==========")
        print("1. Best Overall Recommendation")
        print("2. Recommendation by Genre")
        print("3. Recommendation by OTT Platform")
        print("4. Top 5 Highest-Rated Titles")
        print("5. Search by Title")
        print("6. Show All 100 Titles")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            best_movie = max(movies, key=lambda movie: movie["rating"])
            print("\n------ BEST OVERALL PICK ------")
            print("You should watch:", best_movie["title"])
            print("Platform:", best_movie["platform"])
            print("Genre:", best_movie["genre"])
            print("Sample rating:", best_movie["rating"], "/ 5")

        elif choice == "2":
            genres = []

            for movie in movies:
                if movie["genre"] not in genres:
                    genres.append(movie["genre"])

            print("Available genres:", ", ".join(genres))
            wanted_genre = input("Enter the genre you like: ").lower()

            selected = []
            for movie in movies:
                if movie["genre"].lower() == wanted_genre:
                    selected.append(movie)

            show_recommendations(selected)

        elif choice == "3":
            platforms = []

            for movie in movies:
                if movie["platform"] not in platforms:
                    platforms.append(movie["platform"])

            print("Available platforms:", ", ".join(platforms))
            wanted_platform = input("Enter your OTT platform: ").lower()

            selected = []
            for movie in movies:
                if movie["platform"].lower() == wanted_platform:
                    selected.append(movie)

            show_recommendations(selected)

        elif choice == "4":
            show_recommendations(movies.copy())

        elif choice == "5":
            search_title = input("Enter title or part of title: ").lower()
            selected = []

            for movie in movies:
                if search_title in movie["title"].lower():
                    selected.append(movie)

            show_recommendations(selected)

        elif choice == "6":
            print("\n------ ALL TITLES IN DATASET ------")

            for movie in movies:
                print(
                    movie["title"], "|",
                    movie["platform"], "|",
                    movie["genre"], "|",
                    movie["rating"], "/5"
                )

        elif choice == "7":
            print("Thank you for using OTT Recommendation System!")
            break

        else:
            print("Invalid choice. Please try again.")
```
